```python
import os
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ----------------------------
# 1) Load data
# ----------------------------
FILE_PATH = "DATA/train.xlsx"  # <- change if needed
OUT_DIR = "figures"
os.makedirs(OUT_DIR, exist_ok=True)

df = pd.read_excel(FILE_PATH)

df.head()
```

```python
df.columns
```

```python
import matplotlib.pyplot as plt

# Select the first 5 rows (indexes)
df_5 = df.iloc[200:250]

# Hour columns only
hour_cols = [col for col in df.columns if col.startswith("Hour")]

plt.figure()
for idx, row in df_5.iterrows():
    plt.plot(hour_cols, row[hour_cols], label=f"Index {idx}")

plt.xlabel("Hour of Day")
plt.ylabel("Price")
plt.title("Price Movement Over a Day (5 Indexes)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
